```python
# 기본적인 경사하강법은 진동이 심하거나 수렴이 느린 문제가 있어서
# 여러 가지 옵티마이저가 만들어졌다.

# Momentum(모멘텀)
# 이전 단계의 기울기를 일정 비율로 유지하면서 이동하는 방식.
# 관성처럼 작용해 진동을 줄이고 더 빠르게 수렴하도록 돕는다.

# Adagrad
# 파라미터마다 학습률을 다르게 조정하는 방법.
# 자주 업데이트되는 파라미터는 학습률을 낮추고,
# 드물게 업데이트되는 파라미터는 학습률을 높인다.
# 하지만 시간이 지나면 학습률이 너무 작아지는 단점이 있다.

# RMSProp
# Adagrad의 단점을 해결한 방법으로,
# 최근 기울기의 평균만 고려해 학습률이 지나치게 줄지 않도록 한다.
# 이로써 학습이 더 안정적으로 진행된다.

# Adam
# Momentum과 RMSProp을 합친 형태의 옵티마이저.
# 기울기의 평균(방향 정보)과 제곱 평균(크기 정보)을 모두 사용하여
# 각 파라미터마다 학습률을 자동으로 조정한다.
# 빠르고 안정적인 학습이 가능하다.

# AdamW
# Adam에 정규화(Weight Decay)를 올바르게 적용한 버전.
# Adam의 장점을 유지하면서 과적합을 막고 일반화 성능을 높인다.

# 요약:
# Momentum : 진동 완화, 빠른 수렴
# Adagrad  : 파라미터별 학습률 조정
# RMSProp  : 학습률 감소 문제 해결
# Adam     : 방향성과 적응적 학습률 결합
# AdamW    : 정규화 개선, 더 나은 일반화
```
