# **경사하강법 종류**

**배치 경사하강법 (Batch GD)**

- 전체 데이터로 한 번에 기울기 계산 → 한 스텝 정확하지만 느림, 메모리 큼.

- 장점: 안정적 수렴·진동 적음. 단점: 대용량에서 비현실적.

**확률적 경사하강법 (SGD)**

- 샘플 1개씩 업데이트 → 스텝 빠르고 메모리 적음, 잡음 커서 손실이 요동.

- 장점: 온라인/대용량에 적합, 지역해 탈출 도움. 단점: 수렴 느리고 불안정.

**미니배치 경사하강법 (Mini-batch GD)**

- 보편적 표준. 배치 크기 𝑏
- b (예: 32~1024)로 업데이트.

- 장점: 계산 효율(벡터화) + 노이즈 완화 균형. 단점: 배치 크기 튜닝 필요.

# **옵티마이저**

**Momentum**

- 과거 기울기 지수평균(속도 벡터) 누적 → 관성으로 좁은 골짜기 통과 가속.

- 장점: 수렴 가속, 지그재그 완화. 단점: 모멘텀 계수 튜닝 필요.

- 주요 파라미터: 𝜂, momentum(β, 보통 0.9).

**AdaGrad**

- 좌표별로 과거 제곱그라드를 누적 → 자주 큰 기울기 축은 학습률 자동 감소.

- 장점: 희소 특징(추천·NLP 옛 작업)에 강함. 단점: 누적이 계속 커져 학습률이 급격히 0에 가까워져 후반 둔화.

- 파라미터: 𝜂 (초기), 𝜖

**RMSProp**

- AdaGrad의 “누적 무한증가”를 지수이동평균으로 완화.

- 장점: 비정상/온라인 문제에 안정적, CNN에서 인기. 단점: β, 𝜂 튜닝 필요.

- 파라미터: 𝜂 decay(β≈0.9), 𝜖

**Adam**

- Momentum(1차 모멘트) + RMSProp(2차 모멘트) 결합, 편향보정 포함.

- 장점: 기본값으로도 잘 동작, 초기 수렴 빠름. 단점: 과적합/일반화 논쟁, 학습률 스케줄 중요.

- 파라미터: 𝜂(기본 1e-3), β1=0.9, β2=0.999, 𝜖

**AdamW**

- Adam + 가중감쇠(weight decay) 분리(정규화 항을 업데이트와 분리) → Adam의 L2 정규화 문제 개선.

- 장점: Adam보다 일반화 향상 사례 다수, 현대 딥러닝 기본 선택.

- 파라미터: 𝜂, β1, β2, 𝜖, weight_decay(예: 0.01)
