# 확률적 경사 하강법

훈련한 모델을 버리지 않고 새로운 데이터에 대해서만 조금씩 더 훈련

확률적 경사 하강법은 훈련 세트에서 랜덤하게 하나의 샘플을 선택하여 가파른 경사를 조금 내려감. 그다음 훈련 세트에서 랜덤하게 또 다른 샘플을 하나 선택하여 경사를 조금 내려감. 이런 식으로 전체 샘플을 모두 사용할 때까지 계속함.

# 미니배치 경사 하강법
1개씩 샘플을 선택하지 않고 무작위로 몇 개의 샘플을 선택해서 경사를 내려감

# 배치 경사하강법(경사하강법)
한 번 경사로를 따라 이동하기 위해 전체 샘플을 사용

모델을 학습할 때, 손실함수(loss)가 최소가 되게 모델의 가중치를 조정해야함
-> 그때 사용하는 방법이 **경사하강법**

- 걸음(learning rate) 이 너무 크면 → 아래로 내려가다 튕겨나감

- 너무 작으면 → 끝없이 느림

- 어떤 축은 가파르고, 어떤 축은 완만해서 → 지그재그 진동

- 손실곡선이 복잡할 땐 → 지역 최소(local minima) 에 갇히기도 함



# 옵티마이저란?
경사하강법을 더 효율적이고 안정적으로 만드는 알고리즘

**속도 방향 보폭 **등을 자동으로 조절해줌

# Momentum

이전에 이동한 방향을 기억해서, 그 방향으로 관성(속도)을 유지하면서 나아가는 방법

**해결하려는 문제**
- 경사가 한쪽은 크고, 다른 한쪽은 작은 경우
- 속도를 누적해서 진동을 줄이고 더 빠르게 수렴

# Adagrad

가중치마다 다른 학슙률을 적용해서, 자주 업데이트되는 건 작게, 드물게 업데이트되는건 크게 바꿔주는 방식

**해결하려는 문제**
- 희소 데이터에서 어떤 파라미터는 자주, 어떤 건 거의 안 쓰음 -> 학습률을 똑같이 주면 비효율

# RMSProp

Adagrad의 단점을 보완해서, 최근 기울기만 고려하도록 만든 방법

**해결하려는 문제**
- Adagrad의 학습률이 점점 0에 가까워지는 문제


#Adam

Momentum+RMSProp의 장점을 결합

**해결하려는 문제**
- RMSProp은 진동을 줄이지만 방향감각이 부족
- Momentum은 방향은 좋지만 학습률이 고정
->두 개를 합치면 빠르고 안정적인 학습

# AdamW

Adam의 변형 버전으로, 정규화를 제대로 적용함

**해결하려는 문제**
- Adam은 Weight Decay가 잘못 작동함
- 그 결과 과적합 문제가 생김
