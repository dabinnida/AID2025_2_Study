```python
1. Momentum은 관성 개념을 적용한 옵티마이저다.
이전 단계의 기울기를 일정 비율로 유지하면서 현재 기울기와 결합해 파라미터를 갱신한다. 이는 경사가 급한 방향에서 진동이 심해지는 문제와 평평한 지역에서 수렴이 느려지는 문제를 해결하기 위해 만들어졌다. 물리학의 운동량 개념을 도입해 진동을 줄이고 빠른 수렴을 유도한다.
2. Adagrad는 각 파라미터마다 적응적인 학습률을 부여하는 옵티마이저다.
자주 갱신되는 파라미터의 학습률은 작게, 드물게 갱신되는 파라미터의 학습률은 크게 조정한다. 하나의 전역 학습률만 사용하는 SGD가 희소 데이터에서 비효율적인 문제를 해결하기 위해 제안되었다. 과거의 기울기 제곱을 누적하여 학습률을 자동 조정하지만, 누적값이 계속 커져 학습률이 지나치게 작아지는 단점이 있다.
3. RMSProp은 Adagrad의 학습률이 계속 감소하는 문제를 해결하기 위해 제안된 옵티마이저다.
기울기의 제곱 누적합을 단순 누적이 아닌 지수이동평균으로 관리해 오래된 정보의 영향을 줄인다. 이를 통해 최근 기울기에 더 큰 가중치를 두어 학습률이 안정적으로 유지되도록 한다.
4.Adam은 Momentum과 RMSProp을 결합한 옵티마이저다.
기울기의 1차 모멘트(평균)와 2차 모멘트(분산)를 동시에 추적해 빠르고 안정적인 학습을 수행한다. 단순 SGD나 Momentum, RMSProp에서 발생하는 느린 수렴과 불안정한 갱신 문제를 해결하기 위해 만들어졌다. 학습 초기에 편향이 생기는 문제를 보정하기 위해 바이어스 교정 기법을 사용한다.
5.AdamW는 Adam의 변형 버전으로, Weight Decay를 모멘텀 업데이트와 분리한 옵티마이저다.
 Adam은 L2 정규화 항을 제대로 적용하지 못해 일반화 성능이 떨어지는 문제를 보였다. AdamW는 가중치 감쇠를 파라미터 업데이트 식에서 직접 분리 적용함으로써, 과적합을 줄이고 일반화 성능을 개선한다.
```
